Ask the user for a number. Print its times table from 1 to 10 with a for loop and range().

```python
num = int(input("Which table? "))

for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")

# 5 x 1 = 5
# 5 x 2 = 10
# ...
# 5 x 10 = 50

# Your job:
# 1. ask "Up to?" and use it in range()
# 2. add up all the answers and print the total
```

```python
num = int(input("Which table? "))
upto = int(input("Up to? "))

total = 0

for i in range(1, upto + 1):
    answer = num * i
    print(f"{num} x {i} = {answer}")
    total += answer

print("Total =", total)
```
