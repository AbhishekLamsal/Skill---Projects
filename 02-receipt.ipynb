## Create 02_receipt.py. Ask for an item name, its price, and the quantity bought. Calculate the subtotal, add 13% VAT, and print a neatly formatted receipt with the item name, quantity, and final total — all using f-strings.

```python
item     = input("Item name: ")
price    = float(input("Price: "))
quantity = int(input("Quantity: "))


subtotal = price * quantity

total = subtotal * 1.13

print(f"         --- Receipt ---        ")

print(f"{'Item Name':<15} {'Price':<10} {'Quantity':<10} {'Final Total':<15}")
print(f"{item:<15} {price:<10} {quantity:<10} {total:<15.2f}")
```
