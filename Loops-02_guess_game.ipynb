```python
Keep a secret number. The user keeps guessing until they get it right. Say Too low or Too high to help them. Count the tries.
```

```python
secret = 7
tries = 0

while True:
    guess = int(input("Guess (1 to 10): "))
    tries += 1

    if guess == secret:
        print(f"Correct! You took {tries} tries")
        break
    elif guess < secret:
        print("Too low")
    else:
        print("Too high")

# Your job:
# 1. give only 3 tries (hint: while tries < 3)
# 2. bonus: import random
#    secret = random.randint(1, 10)
```

```python
import random

secret = random.randint(1, 10)
tries = 0

while tries < 3:
    guess = int(input("Guess (1 to 10): "))
    tries += 1

    if guess == secret:
        print(f"Correct! You took {tries} tries")
        break
    elif guess < secret:
        print("Too low")
    else:
        print("Too high")
else:
    print(f"Out of tries! The number was {secret}")
```
