The Day 08 password checker, now split into small functions. Look at has_number(): return inside the loop stops it as soon as a number is found.

```python
def has_number(password):
    for ch in password:
        if ch.isdigit():
            return True
    return False

def is_strong(password):
    return len(password) >= 8 and has_number(password)

while True:
    password = input("New password: ")
    if is_strong(password):
        print("Strong password. Saved!")
        break
    print("Use 8 or more letters and at least one number")

# Your job: write has_capital(password)
# (hint: ch.isupper()) and use it in is_strong()
```

```python
def remove_duplicates(items):
    return sorted(set(items))


def total_price(cart):
    total = 0

    for price in cart.values():
        total += price

    return total


def common(a, b):
    return list(set(a) & set(b))


def costliest(cart):
    item_name = ""
    highest_price = -1

    for name, price in cart.items():
        if price > highest_price:
            highest_price = price
            item_name = name

    return item_name


names = ["Ram", "Sita", "Ram", "Hari", "Sita"]
print(remove_duplicates(names))

cart = {"rice": 1200, "oil": 350, "sugar": 140}
print(total_price(cart))

list1 = ["Ram", "Sita", "Hari"]
list2 = ["Hari", "Gita", "Ram"]

print(common(list1, list2))
print(costliest(cart))
```
