Turn the Day 08 vowel counter into a function. Then write is_palindrome(): a word that reads the same backwards. [::-1] from Day 03 flips a string.

```python
def count_vowels(text):
    count = 0
    for letter in text.lower():
        if letter in "aeiou":
            count += 1
    return count

def is_palindrome(word):
    word = word.lower()
    return word == word[::-1]

print(count_vowels("I love Nepal"))   # 5
print(is_palindrome("Madam"))         # True
print(is_palindrome("Nepal"))         # False

# Your job:
# 1. write count_spaces(text)
# 2. write first_letters(text): "I love Nepal" gives "ILN"
#    (hint: .split() from Day 04)
```

```python
def count_vowels(text):
    count = 0

    for letter in text.lower():
        if letter in "aeiou":
            count += 1

    return count


def is_palindrome(word):
    word = word.lower()
    return word == word[::-1]


def count_spaces(text):
    count = 0

    for letter in text:
        if letter == " ":
            count += 1

    return count


def first_letters(text):
    result = ""

    for word in text.split():
        result += word[0]

    return result


print(count_vowels("I love Nepal"))      
print(is_palindrome("Madam"))            
print(is_palindrome("Nepal"))            

print(count_spaces("I love Nepal"))      
print(first_letters("I love Nepal"))     
```
