## Create 03_words.py. Ask for two sentences. Split each one into a set of words, then print the common words, all the unique words, and how many of each. Store the two counts in a tuple and unpack them to print.

```python
s1 = input("Sentence 1: ").lower()
s2 = input("Sentence 2: ").lower()

words1 = set(s1.split())
words2 = set(s2.split())

common = words1 & words2
print(f"Common words: {sorted(common)}")

all_words = words1 | words2
print(f"All words: {sorted(all_words)}")

counts = (len(common), len(all_words))

common_count, total_words = counts

print(f"Common word count: {common_count}")
print(f"Total unique words: {total_words}")

only_one = words1 ^ words2
print(f"Words in only one sentence: {sorted(only_one)}")
```
