## Create 03_profile_card.py. Ask for the user's full name and birth year. Clean the name with .strip() and .title(). Calculate their age from the current year. Print a profile card showing their name in UPPERCASE, their age, and whether their name is "long" (more than 10 letters) using len() and a comparison.

```python
from datetime import datetime

full_name = input("Enter your full name: ").strip().title()
birth_year = int(input("Enter your birth year: "))

current_year = datetime.now().year
age = current_year - birth_year

is_long_name = len(full_name) > 10

# is_long_name = len(full_name.replace(" ", "")) > 10

print("\n--- Profile Card ---")
print(f"Name       : {full_name.upper()}")
print(f"Age        : {age}")
print(f"Long Name? : {'Yes' if is_long_name else 'No'}")
```
