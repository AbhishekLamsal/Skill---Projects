## Create 01_number_checker.py. Ask the user for a number. Tell them if it is positive, negative or zero. Then use % (from Day 03) to tell if it is even or odd.

```python
num = int(input("Enter a number: "))

if num > 0:
    print(f"{num} is Positive")
elif num < 0:
    print(f"{num} is Negative")
else:
    print(f"{num} is Zero")

if num % 2 == 0:
    print(f"{num} is even")
else:
    print(f"{num} is odd")

if num > 100:
    print(f"{num} is a Big number")

if num >= 1 and num <= 10:
    print(f"{num} is a Small number")
```
