## Create 01_id_card.py. Ask for name, age and city, and put them in a tuple. Unpack the tuple and print a neat ID card. Then make a tuple of 3 marks and print the highest, the lowest and the total. Finally, change the city using the list trick.

```python
name = input("Name: ")
age = input("Age: ")
city = input("City: ")
student = (name, age, city)
n, a, c = student

print("\n===== ID CARD =====")
print(f"Name: {n}")
print(f"Age : {a}")
print(f"City: {c}")

marks = (70, 85, 90)

print(f"\nMarks: {marks}")
print(f"Highest: {max(marks)}")
print(f"Lowest : {min(marks)}")
print(f"Total   : {sum(marks)}")

student_list = list(student)
student_list[2] = input("\nEnter new city: ")

student = tuple(student_list)

print("\nUpdated Student Tuple:")
print(f"{student}")
```
