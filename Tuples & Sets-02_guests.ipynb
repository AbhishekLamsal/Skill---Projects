## Create 02_guests.py. Two friends type their guest lists (names separated by commas, no spaces). Make each list a set, then print: guests on both lists, all guests, and the total number of guests. Add one guest with add(), and remove one with discard().

```python
ram = set(input("Ram's guests: ").split(","))
sita = set(input("Sita's guests: ").split(","))

print(f"On both lists: {ram & sita}")
print(f"All guests   : {ram | sita}")

print(f"Total guests : {len(ram | sita)}")

print(f"Only Ram invited: {ram - sita}")

print(f"Is Hari invited? {'Hari' in ram}")

ram.add("Mina")
ram.discard("Hari")

print(f"\nRam's updated guests: {ram}")
```
