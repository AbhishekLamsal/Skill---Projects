## Create 02_login.py. Keep usernames and passwords in a dictionary. Ask the user to log in. Use in to check the username, then check the password. Clean the name with .strip() and .lower(), so " RAM " also works.

```python
users = {"ram": "ram123", "sita": "sita456"}
users = {"hari": "hari789"}
name = input("Username: ").strip().lower()

password = input("Password: ")

if name not in users:
    print("User not found")

elif password == "":
    print("Password missing")
elif users[name] == password:
    print(f"Welcome, {name.title()}!")
else:
    print("Wrong password")

users = {"hari": "hari789"}
```
