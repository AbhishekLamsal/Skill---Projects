## Create 03_tea_shop.py. Show the menu dictionary. Ask for an item. If it's on the menu, ask how many and print the total. If not, say sorry. Then give a discount with if / elif / else.

```python
menu = {"tea": 20, "coffee": 50, "momo": 150}
print("Menu:", menu)

item = input("What do you want? ").strip().lower()

if item in menu:
    qty = int(input("How many? "))
    total = menu[item] * qty
    print(f"Total: Rs. {total}")

    if total >= 500:
        discount = total * 0.10
        print("10% discount applied!")
    elif total >= 200:
        discount = total * 0.05
        print("5% discount applied!")
    else:
        discount = 0
        print("No discount.")

    final_bill = total - discount
    print(f"Final Bill: Rs. {final_bill:.2f}")

else:
    print("Sorry, we don't have that.")
```
