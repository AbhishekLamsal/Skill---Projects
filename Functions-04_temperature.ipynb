Write two small functions that return the converted temperature. Use float(input()) from Day 03 so the user can type 36.6 too.

```python
def c_to_f(c):
    return c * 9 / 5 + 32

def f_to_c(f):
    return (f - 32) * 5 / 9

print(c_to_f(100))     # 212.0
print(f_to_c(50))      # 10.0

temp = float(input("Temperature in Celsius: "))
print(f"That is {c_to_f(temp)} F")

# Your job:
# 1. ask: C to F, or F to C? Then call the right one (Day 07)
# 2. write is_hot(c) that returns True above 30
```

```python
def c_to_f(c):
    return c * 9 / 5 + 32

def f_to_c(f):
    return (f - 32) * 5 / 9

def is_hot(c):
    return c > 30

choice = input("C to F or F to C? ").strip().upper()

if choice == "C":
    temp = float(input("Temperature in Celsius: "))
    print(f"That is {c_to_f(temp)} F")

    if is_hot(temp):
        print("It's hot!")
    else:
        print("It's not hot.")

elif choice == "F":
    temp = float(input("Temperature in Fahrenheit: "))
    celsius = f_to_c(temp)

    print(f"That is {celsius} C")

    if is_hot(celsius):
        print("It's hot!")
    else:
        print("It's not hot.")

else:
    print("Please choose C or F")
```
