## Create 03_report_card.py. Ask for a name and three marks. Use zip() to build a marks dictionary, then put it inside a report dictionary. Print the report card, the total, the average and the best subject.

```python
name = input("Student name: ")
m1 = int(input("Math: "))
m2 = int(input("Science: "))
m3 = int(input("English: "))

subjects = ["math", "science", "english"]
marks = dict(zip(subjects, [m1, m2, m3]))
report = {"name": name, "marks": marks}

print("\n===== REPORT CARD =====")
print(f"Name : {report['name']}")
print(f"Marks: {report['marks']}")

total = sum(marks.values())
print(f"Total: {total}")

average = total / len(marks)
print(f"Average: {average}")

best_subject = max(marks, key=marks.get)
print(f"Best Subject: {best_subject}")
```
