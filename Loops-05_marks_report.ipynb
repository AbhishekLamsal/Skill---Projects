Keep student marks in a dictionary. Loop with .items() and print Pass or Fail for each student. Then print the class average.

```python
marks = {"Ram": 78, "Sita": 92, "Hari": 35, "Gita": 64}
total = 0

for name, mark in marks.items():
    if mark >= 40:
        print(f"{name}: {mark} Pass")
    else:
        print(f"{name}: {mark} Fail")
    total += mark

average = total / len(marks)
print(f"Class average: {average}")

# Your job:
# 1. find the topper (hint: top = 0, then check each mark)
# 2. give a grade A / B / C with elif (Day 07)
```

```python
marks = {"Ram": 78, "Sita": 92, "Hari": 35, "Gita": 64}

total = 0
top = 0
topper = ""

for name, mark in marks.items():
    if mark >= 40:
        result = "Pass"
    else:
        result = "Fail"

    if mark >= 80:
        grade = "A"
    elif mark >= 60:
        grade = "B"
    else:
        grade = "C"

    print(f"{name}: {mark} {result} Grade {grade}")

    total += mark

    if mark > top:
        top = mark
        topper = name

average = total / len(marks)

print(f"Class average: {average}")
print(f"Topper: {topper} ({top} marks)")
```
