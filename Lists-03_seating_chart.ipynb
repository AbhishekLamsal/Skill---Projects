## Create 03_seating_chart.py. Start with a nested list — 3 rows of 4 seats each, every seat set to "Empty". Let the user type a name and a row/column number to seat them, print the whole chart row by row, then count how many seats are still empty by adding up row.count("Empty") for each row.

```python
chart = [
    ["Empty", "Empty", "Empty", "Empty"],
    ["Empty", "Empty", "Empty", "Empty"],
    ["Empty", "Empty", "Empty", "Empty"],
]

name = input("Student name: ")
row = int(input("Row (0-2): "))
col = int(input("Column (0-3): "))

chart[row][col] = name

print("\nSeating Chart:")
for r in chart:
    print(r)

empty_seats = 0
for r in chart:
    empty_seats += r.count("Empty")

print("\nEmpty seats:", empty_seats)
```
